```python
import ipyparallel as ipp

# Create an MPI cluster with 8 engines
cluster = ipp.Cluster(engines="mpi", n=4)

# Start and connect to the cluster
rc = cluster.start_and_connect_sync()

# Enable IPython magics for parallel processing
rc[:].activate()
```

```python
%%px --block
import rockverse as rv
import numpy as np
a = rv.coordinate(data=np.sort(np.random.rand(10)), name='p', unit='m')
b = rv.coordinate(data=np.sort(np.random.rand(5)), name='ss', unit='km')
c = rv.coordinate(data=np.sort(np.random.rand(9)), name='aa', unit='s')
```
