# Working with coordinates

## The Coordinate class

The Coordinate class extends ParallelArray to represent one-dimensional, numeric, 
sorted, and unchunked arrays that serve as coordinates for tensor-field-like objects. 
Coordinates must be strictly one-dimensional, stored as a single chunk, and sorted in 
ascending or descending order. This design ensures efficient, consistent, and 
parallel-friendly access to coordinate data across MPI processes.
