# Working with coordinates

Multidimensional parallel arrays can be associated with spatial coordinates to create 
structured representations of data in physical or abstract spaces. This 
is achieved by defining a set of one-dimensional coordinate arrays &mdash; one for 
each dimension &mdash; enabling the mapping of array elements to meaningful positions. 

## The Coordinate class

The `Coordinate` class extends `ParallelArray` to represent one-dimensional, numeric, sorted, 
and unchunked arrays that serve as coordinates for tensor-field-like objects. 
This design ensures efficient, consistent, and parallel-friendly 
access to coordinate data across MPI processes.

### Creating a Coordinate

You can create a Coordinate object from array-like data using the `coordinate()` function:

```python
import numpy as np
import rockverse as rv

coord = rv.coordinate(np.linspace(0, 8, 5))
print(coord)
print(coord[...])
```

Coordinates do not need to be equally spaced, as long as the values are sorted and non-repeated.
The code below will work just as fine:

```python
coord = rv.coordinate(np.sort(np.random.rand(10)))
print(coord)
print(coord[...])
```

### Attributes and methods

As a `ParallelArray`-derived class, the `Coordinate` class inherits all properties and methods.
For example, we can set the main attributes:

```python
coord.name = 'x'
coord.unit = 'm'
coord.latex_name = "$x$"
coord.description = "x-coordinate"

print(coord.get_plot_label())
```

Additionally, there are special attributes and methods to streamline working 
with coordinates in an MPI parallel environment.
For example, you can find the coordinate element closest to a specified value:

```python
value = 0.54
closest_value = coord.closest_value(value)
print(f"Closest coordinate value to {value} is {closest_value}")
coord[...]
```

or the index of that element:

```python
index = coord.closest_index(value)
print(f"Closest index to {value} is {index}, coord[{index}]={coord[index]}")
```

## The CoordinateSet Class

The CoordinateSet class represents a collection of `Coordinate` objects. 
It provides indexed access to individual coordinates by integer index or 
coordinate name, and collective management of coordinate metadata attributes 
across MPI ranks.


### Creating a CoordinateSet

The coordinate set is automatically created when constructing tensor field-like 
objects, as we will see later. Here, we'll manually create one by passing 
individual coordinates to the class constructor, allowing us to explore its 
features and capabilities more closely:

```python
coord_x = rv.coordinate(np.linspace(0, 10, 11), name='x', unit='m')
coord_y = rv.coordinate(np.linspace(0, 5, 6), name='y', unit='m')

coords = rv.core.CoordinateSet(coord_x, coord_y)
```

### Accessing Coordinates

Access coordinates by index or name:


```python
# By index:
print(f"{coords[0].get_plot_label()}: {coords[0][...]}")
```

```python
# By name (if assigned)
print(f"{coords['y'].get_plot_label()}: {coords['y'][...]}")
```

### Retrieving Metadata

Get tuples of coordinate names, units, descriptions, LaTeX representations, or plot labels:


```python
print('names:', coords.names)
print('units:', coords.units)
print('labels:', coords.get_plot_labels())
```

See the [API documentation](../../../api/core/coordinateset.rst) for a comprehensive list on the `CoordinateSet` attributes and methods.
